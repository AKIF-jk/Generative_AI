```python
import wandb
import os

# Login using environment variable WANDB_API_KEY
# Make sure to set this in Colab Secrets
wandb.login()
print("✓ W&B login successful!")
```

```python
# Initialize a test run
run = wandb.init(
    project="GenAI-Assignment1",
    name="validation-test",
    config={
        "test": True,
        "framework": "PyTorch",
        "dataset": "Oxford-IIIT Pet"
    }
)

print(f"✓ W&B run initialized!")
print(f"Project: GenAI-Assignment1")
print(f"Run URL: {wandb.run.url}")
```

```python
# Test logging metrics
import random

for i in range(10):
    wandb.log({
        "train/loss": random.random(),
        "val/loss": random.random(),
        "val/ssim": 0.5 + random.random() * 0.5,
        "epoch": i
    })

print("✓ Metrics logging successful!")
```

```python
# Test image logging
import numpy as np

# Create a dummy image
dummy_image = np.random.randint(0, 255, (128, 128, 3), dtype=np.uint8)

# Log the image
wandb.log({"test_image": wandb.Image(dummy_image)})

print("✓ Image logging successful!")
```

```python
# Test table logging
table = wandb.Table(columns=["input", "output", "score"])

for i in range(5):
    table.add_data(
        wandb.Image(np.random.randint(0, 255, (64, 64, 3), dtype=np.uint8)),
        wandb.Image(np.random.randint(0, 255, (64, 64, 3), dtype=np.uint8)),
        random.random()
    )

wandb.log({"test_table": table})

print("✓ Table logging successful!")
```

```python
# Test model checkpoint logging
import torch
import torch.nn as nn

# Create a simple model
model = nn.Linear(10, 2)
torch.save(model.state_dict(), "test_model.pt")

# Log the model
wandb.log_model("./test_model.pt", "test_model")

print("✓ Model logging successful!")
```

```python
# Finish the run
wandb.finish()

print("\n" + "="*50)
print("✓ ALL W&B VALIDATION TESTS PASSED!")
print("="*50)
print("\nYour W&B setup is ready for the assignment.")
print("Check your dashboard at the URL provided above.")
```
